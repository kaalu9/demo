```python
import random
from dataclasses import dataclass, field
```

```python
@dataclass
class Environment:
    temperature: float = field(default_factory=lambda: random.uniform(18, 32))
    light: float = field(default_factory=lambda: random.uniform(0, 1000))
    motion: bool = field(default_factory=lambda: random.choice([True, False]))
    hour: int = field(default_factory=lambda: random.randint(0, 23))

    def step(self):
        self.temperature += random.uniform(-1.5, 1.5)
        self.light = max(0, min(1000, self.light + random.uniform(-150, 150)))
        self.motion = random.choice([True, False])
        self.hour = (self.hour + 1) % 24

    def sense(self):
        return {
            "temperature": round(self.temperature, 1),
            "light": round(self.light, 1),
            "motion": self.motion,
            "hour": self.hour
        }
```

```python
class SmartHomeAgent:
    def __init__(self):
        self.ac = False
        self.heater = False
        self.lights = False

    def act(self, data):
        temp = data["temperature"]
        light = data["light"]
        motion = data["motion"]
        hour = data["hour"]

        night = hour < 6 or hour >= 20

        if temp > 26:
            self.ac, self.heater = True, False
        elif temp < 20:
            self.ac, self.heater = False, True
        else:
            self.ac, self.heater = False, False

        self.lights = motion and (light < 300 or night)

        return (
            "AC:", self.ac,
            "| Heater:", self.heater,
            "| Lights:", self.lights
        )
```

```python
env = Environment()
agent = SmartHomeAgent()
```

```python
for step in range(1, 9):
    data = env.sense()
    agent.act(data)

    print(f"\nStep {step}")
    print("-" * 35)
    print(f"Time        : {data['hour']:02d}:00")
    print(f"Temperature : {data['temperature']} °C")
    print(f"Light level : {data['light']} lux")
    print(f"Motion      : {'Present' if data['motion'] else 'Absent'}")
    print(f"AC          : {'ON' if agent.ac else 'OFF'}")
    print(f"Heater      : {'ON' if agent.heater else 'OFF'}")
    print(f"Lights      : {'ON' if agent.lights else 'OFF'}")

    env.step()
```
